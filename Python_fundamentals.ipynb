```python
#emp_list = [('Ram',10000),('rajesh',12000),('Gopi',14000),('Ankit',11000),('Deepak',13000)]
def average_salary(emp_list):
  l =[]
  n = len(emp_list)
  for i in emp_list:
    for j in i: # for i,j in emp_list:  # unpacking
      if type(j) == int:
       l.append(j)
  avg = sum(l)/n
  return avg

def list_of_emp(emp_list,avg_salary):
  l =[]
  for i in emp_list:
    for j in i: # for i, j in emp_list:
      if type(j) == int:
        if j > avg_salary:
          l.append(i)  #l.append((i,j))
  return l
emp_list = [('Ram',10000),('rajesh',12000),('Gopi',14000),('Ankit',11000),('Deepak',13000)]
avg_salary = average_salary(emp_list)
print(avg_salary)
print(list_of_emp(emp_list,avg_salary))

```

```python
# write a function which return the sum of the numbers in the list and implement above one using unpacking method
def average(emp_list):
  number = len(emp_list)
  sum = 0
  for i,j in emp_list:
    sum+=j
  avg = sum/number
  return avg

def max_list(emp_list, avg):
  l = []
  for i, j in emp_list:
    if j > avg:
      l.append((i,j))
  return l
emp_list = [('Ram',10000),('rajesh',12000),('Gopi',14000),('Ankit',11000),('Deepak',13000)]
avg_salary = average(emp_list)
print(avg_salary)
print(max_list(emp_list,avg_salary))

```

```python
# Scope of variables

# Global
# can use any where in the prgm

# local
# can use only inside the prgm
# to convert local to global - we use global <local_variable>

# non-local
# keyword which helps to modify outer variable inside the inner function without creating a new localvariable
# used in nested functions
```

```python
# Non local variable

# function without nonlocal
def outer():
    count = 0
    def inner():
      count = 1
      print("inner count : ", count)
    inner()
    print("outer count: ",count)
outer()
# In this example, the inner function tries to increment count,
# but it fails to change the outer variable.
# By adding nonlocal, you tell Python to look outside the current function
# and modify the variable in the parent function.

# function with non local
def outer():
    count = 0
    def inner():
      nonlocal count
      count+=1
      print("inner count : ", count)
    inner()
    print("outer count: ",count)
outer()
```

```python
# lambda function
# syn: lambda arguments: expression
# Nameless function, one line, anonymous function
# to reuse the function , we need to use the variable
# can take any number of arguments but only one expression
# type(lambda_variable) is <function>
# best to use for simple operations
# Supports if-else, nested conditions also
# Supports multiple arguments also
# to perform simple function or lambda function on data collection / iterable use map(),filter(), reduce() functions
```

```python
print(lambda a: a ** (1/2)) # prints the memory address,so assign to the value
square_root = lambda a: a ** (1/2)
print(square_root(2))
print(square_root(9))
# multiple argments
sum_it = lambda a,b : a+b
print(sum_it(4,5))
# with condition
# max = lambda x : x>1000
# with if-else condition - error if only if condition is used, need to use else statements for sure
even_numbers = lambda x : "Even" if x%2==0  else "Odd"
print(even_numbers(666))
print(even_numbers(455))
# with nested conditions
check = lambda x: "positive" if x>0 else "negative" if x<0 else "Zero"
print(check(45))
print(check(-45))
print(check(0))
```

```python
# Difference between error and exception

# Errors:
# Detected before the execution of program
# Examples: Syntax error, Indentation error
# Caused program termination

# Exceptions:
# Occur during execution and can be handled using constructs such as try, except,finally or raise
# program can continue after handling  the exception
# Examples : ZeroDivisionError, IndexError, TypeErro
```

```python
def inverse(n):
  return 1/n
try:
  value =int(input("ente any number to find inverse of it: "))
  print(inverse(value))
except ZeroDivisionError as zde:
  print("cannot perform inverse for zero")
  print("try with other number")

except Exception as er:
  print("Captured otherthan excepted condition: ",er)

else:
  print("given input is validated successfully")

finally:
  print("End of the program")
```

```python
# Handling the multiple exceptions in one block
names = ("swathi","hari","ammu")
def sisters(index):
  try :
    return names[index]
  except (TypeError, IndexError) as std_err:
    return f"Error : {std_err}" # if no return statement it will return none when except block is executed
  except Exception as e:
    print("Otherthan excepted error: ",e)
print(sisters(0))
print(sisters(4))
print(sisters("name"))
```

```python
# Regular expression --
# Regular expressions are patterns ised to serach, match and manipulate text.
# In python, regex functionality is provided by build-in "re" module

# Practical Applications - Email validation, Phone number extraction, Password validation, Log file analysis, Data cleaning, Search and replace operations

# Regular Expression Operations / Uses:
# Data validation - check whether input follows a required format [Eg: Emailaddress, phone number, passwrods, ZIP/PIN code]
# Data Extraction - Extract specific information from test [Extract IP from document, Extract errors from logfile]
# Search text - Find words or patterns inside large amounts of text
# Replace text - Automatically replace matching text
# Split text - Split data using multiple seperators
# Log file analysis - Developers and system administrators use regex to extract erros, IP addresses, timestamps ete]
# Web Scraping and Data cleaning - Extract useful information from web pages or clean messy data

# Regex pattersn are usually written ad raw strings , to avoid escaping backslashes twice
# r"\d+" ---------> r"\\d+"
```

```python
# re.findall(search_pattern,data)
# returns all the occurrences of given pattern in list format
# not found any , return empty list
# re.IGNORECASE - case insensitive matching
```

```python
import re
def finding(data, pattern, ignorecase=False):
  if ignorecase:
    result = re.findall(pattern,data,re.IGNORECASE)
  else:
    result = re.findall(pattern,data)
  return result
data = "rain in spain"
pattern = "ai"
print(finding(data, pattern))
print(len(finding(data,pattern)))

pattern1 = "AI"
print(finding(data, pattern1)) # this will return empty list because no matching with AI
print(finding(data, pattern1, ignorecase=True)) # this will return list of matching pattern of AI because ignorecase case passed is True
```

```python
# re.sub(search_pattern, new_value_to_replace, data)
# used to replace text that matches a given regex pattern
# re.sub(patter, replacement, string) --- pattern[regex pattern to serach for], replacement[text or replace matches with], string[Original data]
```

```python
import re
def replacing(data, search_pattern):
  result = re.sub(search_pattern, "", data)
  return result
data = "Spec#1234^&*%@ABC"
search_pattern = r"[^A-Za-z0-9]"
print("Before Cleaning: ",data)
result = replacing(data, search_pattern)
print("After data Cleaning: ", result)
```

```python
# Rearranging the date using capture groups
import re
text = "2026-06-11"
pattern = r"(\d{4})-(\d{2})-(\d{2})"
groups  = re.search(r"(\d{4})-(\d{2})-(\d{2})", text)
print(groups.group(1))   # number1 == \1
print(groups.group(2))   # number2 == \2
print(groups.group(3))   # number3 == \3
re.sub(pattern,r"\3/\2/\1",text)  # using group method setting the date
```

```python
# re.split(pattern, string, maxsplit=0)
# pattern - regex that acts as a seperator
# string - text to split
# maxsplit - maximum number of splits [Optional]

# Use string.split(",") ,method when the seperator is simple and use re.split(r"[,.\s]") when seperators are vary / multiple
```

```python
import re
text = "apple,banana,orange.cherry guava"
# Split on single seperator
result = re.split(r",", text)
print(result)  #['apple', ' banana ', ' orange.cherry']
# Split on multiple seperators
result1 = re.split(r"[,.\s]",text) #[] -- represents match one of the these
print(result1)
# Using maxsplit
res3 = re.split(r",", text, maxsplit=1)
print(res3)
# Capturing groups in re.split
result4 = re.split(r"([,.\s])",text)  # if we use parenthesis in the pattern, the matched separators are also included in the result
print(result4)
```

```python
# re.search(pattern, string)
# used to find the first occurance of a pattern anywhere in the given data
# returns a match object if found
# returns None if no match find
# to access the matched text from re.search() use group()
# group(0) - entire match
# group(1) - first capture grp .....
# group() returns the matched text
# to find the position of string use start(), end(), span()
# Use re.search() when we want to extract the first occurrence of a pattern from text , such as email, phne number, date, ID or keyword
```

```python
import re
text = "I am Hari priya, studying Python"
res = re.search("Python",text)
print(res) # returns the object
if res:
  print("found")
# print the matched text
matched_text = res.group()
print("matched text expecting ", matched_text)
# printing the postions
print(res.start())
print(res.end())
print(res.span())
# what if the match is not found
invalid = re.search("prince",text)
print(invalid)
```

```python
# re.finditer(pattern, string)
# finds all the matches in a string, just like re.findall()
# but instead of returning a list of strings, it returns an iterator of match objects
# this is useful because each match object contains extra information such as
# the matched text, start position, end position, capture groups
# useful when we want all matches and their positions

# difference between findall() and finditer()
# findall() returns only matched strings and finditer() returns match objects with more details
# use findall() when only matched text is needed
# use finditer() when need the matched text + information about where it was found or access to capture groups
```

```python
import re
text = "apple banana cherry apple"
matches = re.finditer("apple", text)
print(matches)
for match in matches:
  print(match)
  print(match.group())
  print(match.span())
```

```python
# re.compile(pattern)
# pattern ----> common pattern
# used to compile a regex pattern into a regex object that can be reused multiple times
# Instead of writing the same pattern repeatedly, you compile it once and use it many times
# Advantages --> reuses the same pattern, cleaner code
# Good when the same regex is used many times
# Not necessary if we are using the pattern only once
```

```python
import re
text = "CP 500 SP 700 PROFIT 200"
pattern = re.compile(r"\d+")
print("seraching with the pattern: ", pattern.search(text).group())
print("finding all the possible patterns: ",pattern.findall(text))
```

```python
# fullmatch()
# checks whether the entire string matches the regex pattern
# returns None if entire text doesnot matches the regex
# search() - looks anywhere, match() - looks only at the start but fullmatch() - looks entire string
# fullmatch() commonly used for email validation, phone numbers, PIN codes, Usernames, Password rules
# fullmatch() is equivalent to using ^[start of the string]and $[end of string]
```

```python
import re
pattern = re.compile(r"\d")
text = "123abc"
print(pattern.search(text))
print(pattern.match(text))
print(pattern.fullmatch(text))
```

```python
# Meta Characters used in re module
# abc...  - letters
# 123...  - digits
# \d - any digit
# \D - any non digit character
# \s - any whitespaces
# \S - any non-whitespace
# \w - any alphanumeric character
# \W - any non-aplphanumeric character
# . - Any character
# \. - consider . as period
# ^ - start of string
# $ - End of string
# * - 0 or more repetitions
# + - 1 or more repetitions
# ? - 0 or 1 [Optional character]
# [abc] - character class - means only a,b,c
# [^abc] - Negated character class - means not a,b, or c
# [a-z] - characters from a to z
# [0-9] - numbers from 0 to 9
# | - OR (abc|edf) - can match abc or edf
# () - group/capture
# (a(bc)) - capture subgroup
# (.*) - capture all
# ^...$ - starts and ends
# {m} - exactly m repetitions
# {m,n} - m to n repetitions
```

```python
# email validation is the best example which cover all the metacharacters
import re
text = input("enter a email to check valid format or not: ")
pattern = r"[a-zA-Z0-9._]+@[a-zA-Z]+\.(com|org|in|net)"
if re.fullmatch(pattern, text):
  print("VALID EMAIL")
else:
  print("INVALID EMAIL")
```
